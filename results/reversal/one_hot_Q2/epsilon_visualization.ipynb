```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

#sns.set_style("whitegrid")
#plt.rcParams.update({'font.size': 12, 'pdf.fonttype': 42})

epsilons_short = np.load('epsilons.npy')
epsilons_long = np.load('epsilons_long.npy')
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(epsilons_short, label='Short (L=40)', kde=True, color='C0', alpha=0.5, stat='density')

plt.title('Distribution of $\\epsilon$', fontsize=16)
plt.xlabel('$\\epsilon$')
plt.ylabel('Density')
plt.legend()
plt.savefig('epsilon_distribution_comparison_short.pdf', bbox_inches='tight')
plt.show()


plt.figure(figsize=(10, 6))
sns.histplot(epsilons_long, label='Long (L=500)', kde=True, color='C1', alpha=0.5, stat='density')
plt.title('Distribution of $\\epsilon$', fontsize=16)
plt.xlabel('$\\epsilon$')
plt.ylabel('Density')
plt.legend()
plt.savefig('epsilon_distribution_comparison_long.pdf', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(15, 6))
L = 40
plt.plot(epsilons_short, 'o-', color='C0', linewidth=1.5, markersize=4)

plt.axvspan(0, L-1, alpha=0.1, color='blue', label='PUSH')
plt.axvspan(L+1, 2*L, alpha=0.1, color='orange', label='POP')

plt.title(f'$\\epsilon_t$ over Time (L={L})', fontsize=16)
plt.xlabel('Sequence Step')
plt.ylabel('$\\epsilon$')
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.savefig('epsilon_time_short_plus.pdf', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(15, 6))
L = 500
plt.plot(epsilons_long, 'o-', color='C0', linewidth=1.5, markersize=4)

plt.axvspan(0, L-1, alpha=0.1, color='blue', label='PUSH')
plt.axvspan(L+1, 2*L, alpha=0.1, color='orange', label='POP')

plt.title(f'$\\epsilon_t$ over Time (L={L})', fontsize=16)
plt.xlabel('Sequence Step')
plt.axvline(x=60, color='red', linestyle='--', label='Training Horizon (L=60)')
plt.ylabel('$\\epsilon$')
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.savefig('epsilon_time_long_plus.pdf', bbox_inches='tight')
plt.show()
```
